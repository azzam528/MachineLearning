```python
df.replace(-200, np.nan, inplace=True)
```

```python
ringkasan_missing = pd.DataFrame({
    "Jumlah Missing": df.isnull().sum(),
    "Persentase Missing (%)":
        (df.isnull().sum() / len(df) * 100).round(2)
})

display(
    ringkasan_missing[
        ringkasan_missing["Jumlah Missing"] > 0
    ].sort_values(
        "Persentase Missing (%)",
        ascending=False
    )
)
```

Dataset telah berhasil dimuat dan diperiksa struktur serta dimensinya. Nilai -200
ditemukan pada beberapa kolom numerik. Berdasarkan dokumentasi dataset, nilai
tersebut bukan hasil pengukuran yang valid, melainkan penanda kegagalan sensor.
Oleh karena itu, seluruh nilai -200 diganti menjadi np.nan agar tidak ikut dihitung
dalam statistik dan analisis.

```python
# Memilih semua kolom numerik
numeric_cols = df.select_dtypes(include='number').columns

# Mengisi missing value dengan interpolasi linear
df[numeric_cols] = df[numeric_cols].interpolate(method='linear', limit_direction='both')
```

```python
df.isnull().sum()
```

```python
df["Time"].head(10)
```

```python
#membuat Kolom Jam
df["Jam"] = pd.to_datetime(df["Time"], format="%H:%M:%S").dt.hour
```

```python
#memeriksa hasil
df[["Time", "Jam"]].head(10)
```

```python
#rata rata co perjam
rata_co_per_jam = (
    df.groupby("Jam")["CO(GT)"]
    .mean()
    .reset_index()
)

rata_co_per_jam
```

```python
#Rata2 CO Terendah
jam_terendah = rata_co_per_jam.idxmin()

nilai_terendah = rata_co_per_jam.min()

print("Jam terendah :", jam_terendah)
print("Rata-rata CO :", nilai_terendah)
```

```python
#Rata2 CO Tertinggi
jam_tertinggi = rata_co_per_jam.idxmax()

nilai_tertinggi = rata_co_per_jam.max()

print("Jam tertinggi :", jam_tertinggi)
print("Rata-rata CO :", nilai_tertinggi)
```
